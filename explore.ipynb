```python
import sqlite3
import os

db_path = "novelty_cache.sqlite3"

print("File exists:", os.path.exists(db_path))
print("Absolute path:", os.path.abspath(db_path))
print("File size:", os.path.getsize(db_path) if os.path.exists(db_path) else "N/A")

conn = sqlite3.connect(db_path)

print(conn.execute("PRAGMA database_list").fetchall())

print(conn.execute("""
    SELECT type, name, sql
    FROM sqlite_master
""").fetchall())

conn.close()
```
